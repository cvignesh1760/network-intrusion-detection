# Network Intrusion Detection: Exploratory Data Analysis (EDA)

## 1. Challenge Overview
Traditional intrusion detection systems often struggle with unfamiliar network traffic behaviors. This challenge focuses on robust multiclass classification into 4 target categories:
- **Benign**: Normal legitimate network operations.
- **Availability**: Denial of Service (DoS/DDoS) flood attacks.
- **Reconnaissance**: Network port scanning and service discovery.
- **Spoofing**: Packet header manipulation and identity spoofing.

**Evaluation Metric**: Macro-Averaged F1 Score across all four classes.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Set paths
DATA_DIR = Path("../data")
train = pd.read_csv(DATA_DIR / "train.csv")
test = pd.read_csv(DATA_DIR / "test.csv")

print(f"Train shape: {train.shape}")
print(f"Test shape:  {test.shape}")
```

## 2. Target Class Distribution & Imbalance Analysis
In competitive machine learning, analyzing class proportions is critical when optimizing Macro F1.

```python
counts = train["Target"].value_counts()
percentages = train["Target"].value_counts(normalize=True) * 100
dist_df = pd.DataFrame({"Count": counts, "Percentage (%)": percentages.round(2)})
print(dist_df)
```

## 3. Investigating Behavioral Shift Between Train & Test
The competition problem specifically notes that evaluation data includes separated network traffic behavior profiles.
Let us examine destination ports across train and test sets.

```python
train_ports = set(train["id.resp_p"].unique())
test_ports = set(test["id.resp_p"].unique())
print(f"Unique destination ports in train: {len(train_ports)}")
print(f"Unique destination ports in test:  {len(test_ports)}")
print(f"Test ports never seen in train:    {len(test_ports - train_ports)}")
print("\nTop 5 Destination Ports in Test:")
print(test["id.resp_p"].value_counts().head(5))
```

## 4. Traffic Flow Characteristics by Class
Rather than memorizing raw port IDs, models should rely on behavioral flow dynamics.

```python
behavior_cols = [
    "flow_duration", "fwd_pkts_tot", "bwd_pkts_tot", "flow_pkts_per_sec",
    "flow_SYN_flag_count", "flow_RST_flag_count", "flow_ACK_flag_count",
    "fwd_pkts_payload.tot", "bwd_pkts_payload.tot", "fwd_init_window_size"
]
print(train.groupby("Target")[behavior_cols].median().T)
```
